```python
# path = r'C:\Users\teo\OneDrive - McGill University\Research stuff\my_images\abberior_images\20241009\AT1R_G13_converted_tiffs\*.tiff'
path = r'C:\Users\teo\OneDrive - McGill University\Research stuff\my_images\abberior_new_FPs\20250207\at1r-gq_converted_tiffs\*.tiff'
df = {
    'date':[],
    'cell':[],
    'name':[],
    'bin':[],
    'path':[],

}

ims = glob.glob(path)
for im in ims:
    df['path'].append(im)
    df['date'].append(im[84:92])
    cell = im[93:110]
    print(cell)
    x = re.search('_', cell).span()[0] + 0 # this is usually where i have to edit things
    cell = cell[:x]
    print(cell)   # use this to check if we have the correct offset for cell
    df['cell'].append(cell)
    ch1 = re.search('405.tiff', im)
    ch2 = re.search('561.tiff', im)
    ch3 = re.search('561d.tiff', im)
    if ch1 is not None:
        df['bin'].append(1)
        bin_x = ch1.span()[0]
        namestart = re.search(r'\\', im[::-1]).span()[0]
        name = im[-namestart:bin_x]
        df['name'].append(name)
    elif ch2 is not None:
        df['bin'].append(2)
        bin_x = ch2.span()[0]
        namestart = re.search(r'\\', im[::-1]).span()[0]
        name = im[-namestart:bin_x]
        df['name'].append(name)
    elif ch3 is not None:
        df['bin'].append(3)
        bin_x = ch3.span()[0]
        namestart = re.search(r'\\', im[::-1]).span()[0]
        name = im[-namestart:bin_x]
        df['name'].append(name)
    else:
        df['bin'].append('error')

print(len(df['path']))
print(len(df['date']))
print(len(df['cell']))
print(len(df['bin']))
print(len(df['name']))
```

```python
gooddf = pd.DataFrame(df)
gooddf
```

```python
savepath = r'C:\Users\teo\OneDrive - McGill University\Research stuff\my_images\abberior_new_FPs\20250207\at1r-gq_binned_tiffs\\' # edit here
subdf = gooddf[gooddf['date']=='20250207'] # edit here
subdf = subdf[subdf['cell'] == 'at1r-gq']  # edit here

names = pd.unique(subdf['name'])

for name in names[:]:
    combined = subdf[subdf['name'] == name]
    print(combined)
    ch1 = combined.loc[combined['bin'] == 1]['path'].tolist()[0]
    ### -32768 is an important correction to get down to photon count values... for some reason the real values are known but offset
    ch1 = tif.imread(ch1) - 32768 
    ch2 = combined.loc[combined['bin'] == 2]['path'].tolist()[0]
    ch2 = tif.imread(ch2) - 32768
    ch3 = combined.loc[combined['bin'] == 3]['path'].tolist()[0]
    ch3 = tif.imread(ch3) - 32768
    print(ch1.shape)
    if len(ch1.shape) == 2:
        newarr = np.zeros((2, ch1.shape[0], ch1.shape[1]))
        newarr[0] = ch1
        newarr[1] = ch2 + ch3
    else:
        newarr = np.zeros((2, ch1.shape[0], ch1.shape[1], ch1.shape[2]))
        newarr[0] = ch1
        newarr[1] = ch2 + ch3
        newarr.swapaxes(3, 1)
        newarr.swapaxes(3,2)

    imdate = combined['date'].tolist()[0]
    imcell = combined['cell'].tolist()[0]
    imname = combined['name'].tolist()[0]
    print(savepath + f'{imdate}_{imcell}_{imname}.tiff')
    tif.imwrite(savepath + f'\\{imdate}_{imcell}_{imname}.tiff', newarr)


```
